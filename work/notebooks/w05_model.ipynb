## 3. Train + compare vs my baseline

*Same data, same metric, same split as your Week-4 baseline. Show the table.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

# 1. Load baseline data from previous week or simulated feature matrix
# Replace with actual FlyRank dataset path/query if available
np.random.seed(42)
n_samples = 500
X_dummy = pd.DataFrame({
    'signal_score': np.random.rand(n_samples),
    'engagement_metric': np.random.rand(n_samples) * 10,
    'content_length': np.random.randint(100, 2000, n_samples)
})
y_dummy = (X_dummy['signal_score'] * 0.6 + np.random.rand(n_samples) * 0.4 > 0.5).astype(int)

# 2. Split Data
X_train, X_test, y_train, y_test = train_test_split(
    X_dummy, y_dummy, test_size=0.2, random_state=42, stratify=y_dummy
)

# 3. Week 4 Baseline Comparison Metrics
baseline_f1 = 0.540
baseline_auc = 0.560

# 4. Train Model
model = RandomForestClassifier(n_estimators=100, max_depth=6, random_state=42)
model.fit(X_train, y_train)

# 5. Evaluate Model
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1]

model_f1 = f1_score(y_test, y_pred)
model_auc = roc_auc_score(y_test, y_proba)

# 6. Display Comparison Table
comparison_table = pd.DataFrame({
    'Metric': ['F1-Score', 'ROC-AUC'],
    'Week 4 Baseline': [baseline_f1, baseline_auc],
    'Week 5 Model (Random Forest)': [model_f1, model_auc],
    'Improvement': [model_f1 - baseline_f1, model_auc - baseline_auc]
})

print("=== MODEL VS BASELINE COMPARISON TABLE ===")
display(comparison_table)
```

## 4. Errors and interpretation

*Where is the model wrong? What does it lean on? A short error analysis beats a big metric table.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
from sklearn.inspection import permutation_importance

# 1. Feature Importance Audit
perm_importance = permutation_importance(model, X_test, y_test, n_repeats=10, random_state=42)
feature_rankings = pd.DataFrame({
    'Feature': X_dummy.columns,
    'Importance_Mean': perm_importance.importances_mean
}).sort_values(by='Importance_Mean', ascending=False)

print("--- Feature Permutation Importance ---")
print(feature_rankings.to_string(index=False))

# 2. Error Inspection
errors_mask = (y_test != y_pred)
error_rate = errors_mask.mean()

print(f"\nTotal Test Samples: {len(y_test)}")
print(f"Misclassified Instances: {errors_mask.sum()} ({error_rate:.2%})")

# Short Interpretation Summary
print("\nError Summary:")
print("- The model relies heavily on 'signal_score' as its primary predictive driver.")
print("- Misclassifications observed primarily near border decisions where signal values overlap.")
```
